## 3. Try it on the example images

These sample images are included in the repo (`examples/`), so you can see the model in action without uploading anything of your own.

```python
!python detect.py --model "$MODEL_PATH" --source tests/images/ --output results_examples --save-csv
```

```python
import matplotlib.pyplot as plt
import cv2
import glob

for img_path in sorted(glob.glob("results_examples/annotated/*.jpg")):
    img = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 8))
    plt.imshow(img)
    plt.title(img_path)
    plt.axis("off")
    plt.show()
```

## 4. Try it on your own images

Run the cell below and select one or more images to upload.

```python
import os
from google.colab import files

os.makedirs("my_images", exist_ok=True)
uploaded = files.upload()
for fname in uploaded:
    os.rename(fname, f"my_images/{fname}")
print(f"Uploaded {len(uploaded)} image(s) to my_images/")
```

```python
!python detect.py --model "$MODEL_PATH" --source my_images/ --output results_mine --save-csv
```

```python
for img_path in sorted(glob.glob("results_mine/annotated/*.jpg")):
    img = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 8))
    plt.imshow(img)
    plt.title(img_path)
    plt.axis("off")
    plt.show()
```
